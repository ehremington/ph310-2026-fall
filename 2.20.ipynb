```python
import numpy as np
import math as math
import pandas as pd
import matplotlib.pyplot as plt
from sympy import symbols, Integral, integrate
```

```python
vx0 = vy0 = g = 1

vter = 1

R = 2*vx0*vy0/g*(1-4/3*vy0/vter)

x = np.linspace(0, R, 1000)

def y(x, vx0, vy0, g, vter):
    return ((vy0 + vter) / vx0 * x + (vter**2 / g) * np.log(1 - (x * g) / (vx0 * vter)))
```

```python

fig0, ax0 = plt.subplots()
ax0.plot(x, y(x, vx0, vy0, g, vter))
ax0.set_xlabel('horizontal distance')
ax0.set_ylabel('vertical distance')
```

```python
vx0 = vy0 = g = 1

vter = 1

t = np.linspace(0,2,1000)

def x(t, vx0, vy0, g, vter):
    return(vx0*vter/g*(1-np.exp(-t*g/vter)))

def y(t, vx0, vy0, g, vter):
    return((vy0+vter)*vter/g*(1-np.exp(-t*g/vter))-vter*t)

fig1, ax1 = plt.subplots()

for vter in [0.3, 1, 3, 1000]:
    ax1.plot(x(t, vx0, vy0, g, vter), y(t, vx0, vy0, g, vter))

ax1.legend([0.3, 1, 3, 1000])
```
